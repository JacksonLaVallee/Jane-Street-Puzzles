```python
def find_numbers_2(start1, end1, start2, end2):
    results = []
    for i in range(start1, end1 + 1):  # First digit in range 2-5
        for j in range(start2, end2 + 1):  # Second digit in range 3-7
            num = int(f"{i}{j}")
            if all(d != '0' and num % int(d) == 0 for d in str(num)):
                results.append(num)
    return results

def find_numbers_3(start1, end1, start2, end2, start3, end3):
    results = []
    for i in range(start1, end1 + 1):  # First digit in the specified range
        for j in range(start2, end2 + 1):  # Second digit in the specified range
            for k in range(start3, end3 + 1):  # Third digit in the specified range
                num = int(f"{i}{j}{k}")
                if all(d != '0' and num % int(d) == 0 for d in str(num)):
                    results.append(num)
    return results

def find_numbers_5(start1, end1, start2, end2, start3, end3, start4, end4, start5, end5):
    results = []
    for i in range(start1, end1 + 1):  # First digit in the specified range
        for j in range(start2, end2 + 1):  # Second digit in the specified range
            for k in range(start3, end3 + 1):  # Third digit in the specified range
                for l in range(start4, end4 + 1):  # Fourth digit in the specified range
                    for m in range(start5, end5 + 1):  # Fifth digit in the specified range
                        num = int(f"{i}{j}{k}{l}{m}")
                        if all(d != '0' and num % int(d) == 0 for d in str(num)):
                            results.append(num)
    return results

def find_numbers_6(start1, end1, start2, end2, start3, end3, start4, end4, start5, end5, start6, end6):
    results = []
    for i in range(start1, end1 + 1):  # First digit in the specified range
        for j in range(start2, end2 + 1):  # Second digit in the specified range
            for k in range(start3, end3 + 1):  # Third digit in the specified range
                for l in range(start4, end4 + 1):  # Fourth digit in the specified range
                    for m in range(start5, end5 + 1):  # Fifth digit in the specified range
                        for n in range(start6, end6 + 1):  # Sixth digit in the specified range
                            num = int(f"{i}{j}{k}{l}{m}{n}")
                            if all(d != '0' and num % int(d) == 0 for d in str(num)):
                                results.append(num)
    return results

def find_numbers_7(start1, end1, start2, end2, start3, end3, start4, end4, start5, end5, start6, end6, start7, end7):
    results = []
    for i in range(start1, end1 + 1):  # First digit in the specified range
        for j in range(start2, end2 + 1):  # Second digit in the specified range
            for k in range(start3, end3 + 1):  # Third digit in the specified range
                for l in range(start4, end4 + 1):  # Fourth digit in the specified range
                    for m in range(start5, end5 + 1):  # Fifth digit in the specified range
                        for n in range(start6, end6 + 1):  # Sixth digit in the specified range
                            for o in range(start7, end7 + 1):  # Seventh digit in the specified range
                                num = int(f"{i}{j}{k}{l}{m}{n}{o}")
                                if all(d != '0' and num % int(d) == 0 for d in str(num)):
                                    results.append(num)
    return results

def find_numbers(digit_ranges):
    results = []
    def helper(current, idx):
        if idx == len(digit_ranges):
            num = int("".join(map(str, current)))
            if all(d != 0 and num % d == 0 for d in current):
                results.append(num)
            return
        for d in range(digit_ranges[idx][0], digit_ranges[idx][1] + 1):
            helper(current + [d], idx + 1)
    helper([], 0)
    return results

print("2 digit: " + str(find_numbers_2(2,2,1,9)))

print("3 digit: " + str(find_numbers_3(4,9,1,1,6,9)))

print("5 digit: " + str(find_numbers_5(1,3,4,5,4,5,1,1,6,9)))

print("6 digit: " + str(find_numbers([[3,3],[1,2],[4,5],[4,5],[1,3],[6,9]])))

print("7 digit: " + str(find_numbers_7(3,3,6,6,6,9,4,9,4,4,1,1,1,9)))
```

```python
import math

# Initialize a 2-row, 11-column grid with all 2s
arr = [[2 for _ in range(11)] for _ in range(2)]

def validate_row(arr, row_idx):
    result = []
    temp = []
    for val in arr[row_idx]:
        if val == 0:
            if len(temp) >= 2:
                result.append(int("".join(map(str, temp))))
            elif len(temp) == 1:
                return False
            temp = []
        else:
            if val < 1 or val > 9:
                return False
            temp.append(val)
    if len(temp) >= 2:
        result.append(int("".join(map(str, temp))))
    elif len(temp) == 1:
        return False

    if row_idx == 0:
        for val in result:
            root = int(math.isqrt(val))
            if root * root != val:
                return False
    else:
        for val in result:
            product = 1
            for digit in str(val):
                product *= int(digit)
            if product != 20:
                return False
    return True

def is_adjacent_zero(arr, i, j):
    for di, dj in [(0, -1), (0, 1), (-1, 0), (1, 0)]:
        ni, nj = i + di, j + dj
        if 0 <= ni < 2 and 0 <= nj < 11:
            if arr[ni][nj] == 0:
                return True
    return False

def gen_partitions(total):
    for a in range(total + 1):
        for b in range(total - a + 1):
            for c in range(total - a - b + 1):
                d = total - a - b - c
                yield (a, b, c, d)

def solve(arr, i, j):
    if i == 0 and j == 11:
        if not validate_row(arr, 0):
            return
        solve(arr, 1, 0)
        return

    if i == 1 and j == 11:
        if validate_row(arr, 1):
            for row in arr:
                print("".join(str(x) if x != 0 else "." for x in row))
            print()
        return

    if (i, j) == (1, 1):
        for k in range(1, 10):
            if k == 2:
                continue
            arr[i][j] = k
            solve(arr, i, j + 1)
        arr[i][j] = 0
        for displaced in range(1, 10):
            for up, down, left, right in gen_partitions(displaced):
                deltas = [(-1, 0, up), (1, 0, down), (0, -1, left), (0, 1, right)]
                if any(
                    arr[i + di][j + dj] + bump > 9
                    for di, dj, bump in deltas
                    if 0 <= i + di < 2 and 0 <= j + dj < 11
                ):
                    continue
                for di, dj, bump in deltas:
                    ni, nj = i + di, j + dj
                    if 0 <= ni < 2 and 0 <= nj < 11:
                        arr[ni][nj] += bump
                solve(arr, i, j + 1)
                for di, dj, bump in deltas:
                    ni, nj = i + di, j + dj
                    if 0 <= ni < 2 and 0 <= nj < 11:
                        arr[ni][nj] -= bump
        arr[i][j] = 2
        return

    solve(arr, i, j + 1)

    if not is_adjacent_zero(arr, i, j):
        arr[i][j] = 0
        for displaced in range(1, 10):
            for up, down, left, right in gen_partitions(displaced):
                deltas = [(-1, 0, up), (1, 0, down), (0, -1, left), (0, 1, right)]
                if any(
                    arr[i + di][j + dj] + bump > 9
                    for di, dj, bump in deltas
                    if 0 <= i + di < 2 and 0 <= j + dj < 11
                ):
                    continue
                for di, dj, bump in deltas:
                    ni, nj = i + di, j + dj
                    if 0 <= ni < 2 and 0 <= nj < 11:
                        arr[ni][nj] += bump
                solve(arr, i, j + 1)
                for di, dj, bump in deltas:
                    ni, nj = i + di, j + dj
                    if 0 <= ni < 2 and 0 <= nj < 11:
                        arr[ni][nj] -= bump
        arr[i][j] = 2

solve(arr, 0, 0)
```
