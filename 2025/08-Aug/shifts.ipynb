```python
faces = ["flushedface", "droolingface", "cowboyhat", "woozyface", "anguishedface", "poutingcat", "confounded", "squintingface"]
cards = [[4, 5], [6, 9], [9, 4], [5, 8], [7, 5], [1, 8], [5, 6], [8, 6]]
separated_cards = [4, 5, 6, 9, 9, 4, 5, 8, 7, 5, 1, 8, 5, 6, 8, 6]
chip_counts = [1, 0, 22, 23, 7, 12, 10, 11, 23, 6, 16, 0, 23, 0, 14, 25]
```

```python
hint = ""
for i, f in enumerate(faces):
    for v in cards[i]:
        hint += f[v - 1]
print(hint)
```

```python
phrase = ""
for i, c in enumerate(chip_counts):
    shifted = chr((ord(hint[i]) - ord('a') + c) % 26 + ord('a'))
    phrase += shifted
print(phrase)
```
